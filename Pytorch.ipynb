# Pytorch Computer Vision

```python
import torch
from torch import nn
import torchvision #for datastes, model arch, image transformations
from torchvision import datasets
from torchvision.transforms import ToTensor #for transforming images e.g into numbers, augmented images etc


import matplotlib.pyplot as plt # for visualization

print(f"PyTorch version: {torch.__version__}\ntorchvision version: {torchvision.__version__}")
```

```python
#Getting FashionMNIST dataset
#MNIST stands for Modified National Institute of Standards and Technology.

train_data = datasets.FashionMNIST(
    root="data",
    train=True, #get training data
    download=True,
    transform=ToTensor(),
    target_transform=None
)

test_data = datasets.FashionMNIST(
    root="data",
    train=False,
    download=True,
    transform=ToTensor()
)
```

```python
image, label = train_data[0]
image, label
```

```python
image.shape
#[color_channel, height, width]
```

```python
#Checking number of samples
len(train_data.data), len(train_data.targets), len(test_data.data), len(test_data.targets)
```

```python
class_names = train_data.classes
class_names
```

```python
image, label = train_data[0]
print(f"Image shape: {image.shape}")
plt.imshow(image.squeeze())
plt.title(label);
```
